# Lambda Functions

- A lambda function in Python is a small anonymous function (a function without a name).
It is written in one line and used for short, simple operations.



```python
## Syntax

lambda arguments: expression
```

```python
def addition(a,b):
    return a+b
```

```python
addition(2,5)
```

```python
addition =  lambda a,b: a+b
type(addition)
print(addition(5,6))
```

```python
def even(num):
    if num %2 == 0:
        return True
    
even(24)
```

```python
even1= lambda num:num%2 == 0
even1(12)
```

```python
def addition(x,y,z):
    return x+y+z

addition(12,14,15)
```

```python
addition = lambda x,y,z:x+y+z
addition(12,14,15)
```

```python
## map () - applies a function to all items in a list.
number = [1,2,3,4,5]
def square(number):
    return number**2

square(2)
```

```python
list(map(lambda x:x**2,number))
```
