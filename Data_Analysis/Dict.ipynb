```python
## Creating Dictionaries

empty_dict= {}
print(type(empty_dict))
```

```python
empty_dict = dict()
empty_dict
```

```python
Student = {"name":"Ansh Verma","Age":"23","grade":"24"}
print(Student)
print(type(Student))
```

```python
# Single Key is always used
Student = {"name":"Ansh Verma","Age":23,"name":24}
print(Student)
```

```python
## accesssing Dictionary Elements
Student={"name":"Ansh","Age":23,"grade":'A'}
print(Student)
```

```python
## accesssing Dictionary Elements
print(Student['grade'])
print(Student['Age'])

## Accessing using get() method
print(Student.get('grade'))
print(Student.get('last_name'))
print(Student.get('last_name','Not Available'))
```

```python
## Modifying Dictionary Elements
## Dictionary are mutable, so you can add,update or delete elements
print(Student)
```

```python
Student["Age"]=33 ## Update value for the key
print(Student)
Student["Address"]="India" ## Added a new Key and value
print(Student)
```

```python
del Student['grade'] ## delete key and value pair

print(Student)
```

```python
## Dictionary Methods 

keys = Student.keys() ## All the Keys 
print(keys)
values=Student.values() ## All the Values
print(values)

items = Student.items() ## get all key value Pairs
print(items)
```

```python
## Shallow copy
Student_Copy=Student
print(Student)
print(Student_Copy)
```

```python
Student["name"]= "Ansh2"
print(Student)
print(Student_Copy)
```

```python
Student_Copy1 = Student.copy() ## Shallow copy
print(Student_Copy1)
print(Student)
```

```python
Student["name"] = "Ansh3"
print(Student_Copy1)
print(Student)
```
